```python
 from sklearn.ensemble import RandomForestClassifier
```

```python
y = train_csv["Survived"]

features = ["Pclass", "Sex", "SibSp", "Parch"]
X = pd.get_dummies(train_csv[features])
X_test = pd.get_dummies(test_csv[features])


model = RandomForestClassifier(n_estimators=100, max_depth = 5, random_state=1)
model.fit(X,y)
predictions = model.predict(X_test)

output = pd.DataFrame({'PassengerId': test_csv.PassengerId, 'Survived': predictions})

output.to_csv('my_submission.csv', index=False)
print("yoursubmission was sussesfully saved?")
```
